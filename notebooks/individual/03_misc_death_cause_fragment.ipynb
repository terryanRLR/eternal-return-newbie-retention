````python
import seaborn as sns
import matplotlib.pyplot as plt

# 시각화를 위한 데이터 재구조화
plot_death_df = death_stats.melt(id_vars='Death_Category', value_vars=['Win_Rate', 'Top3_Rate'],
                                 var_name='Metric', value_name='Percentage')

plt.figure(figsize=(12, 6))
sns.barplot(data=plot_death_df, x='Death_Category', y='Percentage', hue='Metric', palette='viridis')

plt.title('Win Rate & Top 3 Rate by Death Cause', fontsize=15)
plt.ylabel('Percentage (%)')
plt.xlabel('Death Category')
plt.xticks(rotation=15)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend(title='Metric')

# 수치 표시
for p in plt.gca().patches:
    plt.gca().annotate(f'{p.get_height():.1f}%',
                   (p.get_x() + p.get_width() / 2., p.get_height()),
                   ha = 'center', va = 'center',
                   xytext = (0, 9),
                   textcoords = 'offset points')

plt.tight_layout()
plt.show()
````
